### Reflection Answers

1. **Why is NumPy faster than Python lists?** NumPy stores data in contiguous memory of a fixed type and runs operations using pre-compiled C code across the whole array at once (vectorization), instead of looping element-by-element in the Python interpreter, which has much higher per-element overhead.

2. **Broadcasting example:** Adding a fixed brightness boost to every pixel in an image, e.g. `img_array + 20`, broadcasts the single scalar `20` across every pixel in every channel without writing a loop.

## Step 3: Word Frequency Vector (Bag-of-Words)



### Demonstration: build a word-count vector by hand using NumPy


```
def bag_of_words_vector(tokens, vocabulary):
    # For each word in the vocabulary, count how many times it appears in tokens
    vector = np.array([np.sum(np.array(tokens) == word) for word in vocabulary])
    return vector

vec = bag_of_words_vector(tokens, vocabulary)
print(vocabulary)
print(vec)                                #AP24110011675
```

The code checks each vocabulary word, counts how many times it appears in the text, and stores those counts as a NumPy array.

## Step 4: Comparing Two Sentences (Cosine Similarity)



$$\text{cosine similarity} = \frac{A \cdot B}{\|A\| \times \|B\|}$$

This measures the angle between two vectors, not their length — so it works even if one sentence is longer than another.

### Demonstration


```
sentence_a = "machine learning uses numpy arrays"
sentence_b = "deep learning also uses numpy arrays heavily"

tokens_a = sentence_a.lower().split()
tokens_b = sentence_b.lower().split()

# Build a shared vocabulary from both sentences combined
combined_vocab = np.unique(np.array(tokens_a + tokens_b))

vec_a = bag_of_words_vector(tokens_a, combined_vocab)
vec_b = bag_of_words_vector(tokens_b, combined_vocab)

print("Vector A:", vec_a)
print("Vector B:", vec_b)
                                         #AP24110011675
# Cosine similarity using NumPy dot product and norm
cosine_sim = np.dot(vec_a, vec_b) / (np.linalg.norm(vec_a) * np.linalg.norm(vec_b))
print("Cosine similarity:", cosine_sim)
```

Take two sentences to compare.
Split them into words using .split().
Create a common vocabulary from both sentences.
Convert each sentence into a number vector using word counts.
Calculate cosine similarity between the two vectors.
Print the similarity score to see how similar the sentences are.

### Student Practice

Pick two sentences of your own (write them below) — one pair that should be *similar* in meaning/topic, and one pair that should be *very different*. Compute the cosine similarity for both pairs and confirm your intuition matches the numbers.


```
similar_a = "the weather today is sunny and warm"
similar_b = "today the weather is warm and sunny"

different_a = "the weather today is sunny and warm"
different_b = "i am learning numpy for my exam"

def cosine_similarity(sent1, sent2):
    t1, t2 = sent1.lower().split(), sent2.lower().split()
    vocab = np.unique(np.array(t1 + t2))
    v1, v2 = bag_of_words_vector(t1, vocab), bag_of_words_vector(t2, vocab)
    return np.dot(v1, v2) / (np.linalg.norm(v1) * np.linalg.norm(v2))

print("Similar pair cosine similarity:", cosine_similarity(similar_a, similar_b))
print("Different pair cosine similarity:", cosine_similarity(different_a, different_b))
```

Reflection :

 Cosine similarity measures the angle between two word-count vectors rather than their raw length. When two sentences share many of the same words in similar proportions, their vectors point in nearly the same direction, giving a value close to 1. When sentences share few or no words, their vectors point in very different directions, giving a value close to 0. A cosine similarity of exactly 1 means the two vectors point in exactly the same direction - i.e. the sentences use the same words in the same relative proportions.

# Mini Project: Simple Text Similarity Checker (NLP Track)



```
sentence1 = "numpy makes numerical computing fast and simple"
sentence2 = "numpy helps with fast numerical computations easily"
sentence3 = "i love eating pizza on weekends"

def cosine_similarity_shared(sent1, sent2, vocab):
    t1, t2 = sent1.lower().split(), sent2.lower().split()
    v1, v2 = bag_of_words_vector(t1, vocab), bag_of_words_vector(t2, vocab)
    return np.dot(v1, v2) / (np.linalg.norm(v1) * np.linalg.norm(v2))

all_tokens = sentence1.lower().split() + sentence2.lower().split() + sentence3.lower().split()
shared_vocab = np.unique(np.array(all_tokens))

sim_12 = cosine_similarity_shared(sentence1, sentence2, shared_vocab)
sim_13 = cosine_similarity_shared(sentence1, sentence3, shared_vocab)
sim_23 = cosine_similarity_shared(sentence2, sentence3, shared_vocab)

print("Similarity 1-2:", sim_12)
print("Similarity 1-3:", sim_13)                               #AP24110011675
print("Similarity 2-3:", sim_23)

pairs = {"Sentence 1 & 2": sim_12, "Sentence 1 & 3": sim_13, "Sentence 2 & 3": sim_23}
most_similar = max(pairs, key=pairs.get)
print("\nMost similar pair:", most_similar, "with similarity", pairs[most_similar])
```
