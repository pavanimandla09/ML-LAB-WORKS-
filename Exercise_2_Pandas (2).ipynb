## 1. Importing pandas

Pandas is usually imported with the short alias `pd`. This is the standard convention followed
almost everywhere in the pandas documentation and in real projects.

```python
import pandas as pd
import numpy as np
```

## 2. Creating a Series from an Array

A `Series` is a **one-dimensional** labelled array that can hold any data type. Here we create
one from a NumPy array of characters.

```python
info = np.array(['S','I','V','A'])
a = pd.Series(info)
print(a)
```

## 3. Creating an Empty DataFrame

A `DataFrame` is a 2-D, table like structure with rows and columns. Let's first see what an
empty one looks like.

```python
df = pd.DataFrame()
print(df)
```

## 4. Creating a DataFrame from a List

We can directly pass a Python list to `pd.DataFrame()` to make a single-column DataFrame.

```python
x = ['Siva', 'Rama', 'Krishna']
df = pd.DataFrame(x)
print(df)
```

## 5. Creating a DataFrame from a Dictionary of Lists

Each key in the dictionary becomes a column name, and the corresponding list becomes the
column values.

```python
info = {'ID': [1001, 1002, 1003], 'Department': ['B.Tech', 'M.Tech', 'PhD']}
df = pd.DataFrame(info)   ##AP24110011675
print(df)
```

## 6. Creating a DataFrame from a Dictionary of Series

When the dictionary values are `Series` with different indexes, pandas aligns them and fills
the missing values with `NaN`.

```python
info = {'one': pd.Series([1, 2, 3, 4, 5, 6], index=['a', 'b', 'c', 'd', 'e', 'f']),
        'two': pd.Series([1, 2, 3, 4, 5, 6, 7, 8], index=['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h'])}
d1 = pd.DataFrame(info)
print(d1)
```

## 7. Column Selection

A single column can be picked out using `df['column_name']`. This returns a `Series`.

```python
info = {'one': pd.Series([1, 2, 3, 4, 5, 6], index=['a', 'b', 'c', 'd', 'e', 'f']),
        'two': pd.Series([1, 2, 3, 4, 5, 6, 7, 8], index=['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h'])}
d1 = pd.DataFrame(info)
print(d1['one'])
print(d1['two'])
```

## 8. Column Addition

New columns can be added simply by assigning to a new column name, either from a `Series`
or from an operation on existing columns.

```python
info = {'one': pd.Series([1, 2, 3, 4, 5], index=['a', 'b', 'c', 'd', 'e']),
        'two': pd.Series([1, 2, 3, 4, 5, 6], index=['a', 'b', 'c', 'd', 'e', 'f'])}
df = pd.DataFrame(info)
print(df)

# Add a new column to an existing DataFrame object
print("Add new column by passing series")
df['three'] = pd.Series([20, 40, 60], index=['a', 'b', 'c'])
print(df)

print("Add new column using existing DataFrame columns")
df['four'] = df['one'] + df['three']              #AP24110011675
print(df)
```

## 9. Column Deletion

Columns can be removed with the `del` keyword or the `.pop()` method.

```python
info = {'one': pd.Series([1, 2], index=['a', 'b']),
        'two': pd.Series([1, 2, 3], index=['a', 'b', 'c'])}
df = pd.DataFrame(info)
print("The DataFrame:")
print(df)

# using del function
print("Delete the first column:")
del df['one']
print(df)

# using pop function
print("Delete the another column:")
df.pop('two')
print(df)
```

## 10. Selection by Label (`loc`)

Rows can be selected using their **label** with the `.loc[]` function.

```python
info = {'one': pd.Series([1, 2, 3, 4, 5], index=['a', 'b', 'c', 'd', 'e']),
        'two': pd.Series([1, 2, 3, 4, 5, 6], index=['a', 'b', 'c', 'd', 'e', 'f'])}
df = pd.DataFrame(info)
print("Data frame:")
print(df)
print("b index elements are:")
print(df.loc['b'])
print("e index elements are:")
print(df.loc['e'])
```

## 11. Selection by Integer Location (`iloc`)

Rows can also be selected by their integer position using `.iloc[]`.

```python
info = {'one': pd.Series([1, 2, 3, 4, 5], index=['a', 'b', 'c', 'd', 'e']),
        'two': pd.Series([1, 2, 3, 4, 5, 6], index=['a', 'b', 'c', 'd', 'e', 'f'])}
df = pd.DataFrame(info)
print(df)
print(df.iloc[3])
```

## 12. Slice Rows

Multiple rows can be selected at once using the `:` (slice) operator.

```python
info = {'one': pd.Series([1, 2, 3, 4, 5], index=['a', 'b', 'c', 'd', 'e']),
        'two': pd.Series([1, 2, 3, 4, 5, 6], index=['a', 'b', 'c', 'd', 'e', 'f'])}
df = pd.DataFrame(info)
print(df)                            #AP24110011675
print(df[2:5])
```

## 13. Addition of Rows

New rows can be added by combining two DataFrames using `pd.concat()`
(the older `.append()` function has been removed from newer pandas versions, so `concat`
is used instead).

```python
d = pd.DataFrame({'A': [1, 2, 3], 'B': [4, 5, 6]})
d2 = pd.DataFrame({'A': [7, 8, 9], 'B': [10, 11, 12]})

# Replace append with concat
d = pd.concat([d, d2])
print(d)
```

## 14. Deletion of Rows

Rows can be dropped using the `.drop()` method with the index label. If the label is
duplicated (as happens after a `concat`), all matching rows get removed.

```python
a_info = pd.DataFrame([[4, 5], [6, 7]], columns=['x', 'y'])
b_info = pd.DataFrame([[8, 9], [10, 11]], columns=['x', 'y'])
a_info = pd.concat([a_info, b_info])
print(a_info)

# Drop rows with label 0
a_info = a_info.drop(0)
print("Drop the row with label 0")
print(a_info)
```

## 15. Custom Column Order & Attribute-style Access

We can control the order of the columns while creating the DataFrame, and columns can
also be accessed like attributes, e.g. `df.Emp_ID`.

```python
data = {'Gender': ['F', 'M', 'M'], 'Emp_ID': ['E01', 'E02', 'E03'], 'Age': [25, 27, 25]}

# We want to order the columns, so let's specify the columns parameter
df = pd.DataFrame(data, columns=['Emp_ID', 'Gender', 'Age'])
print(df)              # output all data
print(df.Emp_ID)       # output all Emp_ID (entire column)
print(df.Emp_ID[0])    # output first Emp_Id
```

## 17. `DataFrame.drop_duplicates()`

This function removes duplicate rows from a DataFrame, keeping only the first
occurrence.

```python
emp = {"Name": ["Siva", "Rama", "Krishna", "Siva"],
       "Age": [21, 32, 29, 21]}
info = pd.DataFrame(emp)
print(info)
info = info.drop_duplicates()
print("After removing the duplicates the dataframe is :")
print(info)
```

## 18. Identifying Null Values in a Dataset

`isnull()` returns a boolean DataFrame marking missing (`NaN`) values, and `.sum()` on top
of it gives a count of missing values per column.

```python
info = pd.DataFrame({"Person": ["Siva", "Pavan", "Gopal", "Raghu"],
                      "Age": [27., 29, np.nan, 32]})
print(info)
print(info.count())
print("null values")
print(info.isnull())
print(info.isnull().sum())
```

## 19. Sorting Order

`.sort_index()` sorts the DataFrame by its row index, either ascending (default) or
descending.

```python
info = pd.DataFrame({"Person": ["Siva", "Pavan", "Gopal", "Raghu"],
                      "Age": [27., 29, np.nan, 32]})
print(info)
print("Sorting Order")
print(info.sort_index())
print("Descending Order")
print(info.sort_index(ascending=False))
```

## 20. Aggregation Functions (groupby)

`groupby()` lets us group rows that share a value in one column (here `species`) and then
apply an aggregation function such as `.mean()` or `.sum()` to each group.

```python
df = pd.read_csv('https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv')
print(df)

mean_df = df.groupby('species').mean(numeric_only=True)
print("Mean values for each species:")
print(mean_df)

sum_df = df.groupby('species').sum(numeric_only=True)
print("\nSum of values for each species:")
print(sum_df)                                                                        #AP24110011675
```
