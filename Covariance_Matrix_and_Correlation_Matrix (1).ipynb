##Python Program to Compute Covariance Matrix and Correlation Matrix

```python
import pandas as pd

# Create a dataset
data = {
    'A': [2, 4, 6, 8, 10],
    'B': [1, 3, 5, 7, 9],
    'C': [5, 10, 15, 20, 25]
}
                                               #AP24110011675
df = pd.DataFrame(data)

# Compute covariance matrix
print("Covariance Matrix:")
print(df.cov())

# Compute correlation matrix
print("\nCorrelation Matrix:")
print(df.corr())
```

Import the Pandas library and create a dataset containing multiple attributes. Convert the dataset into a DataFrame using pd.DataFrame(). Use df.cov() to calculate the covariance matrix and df.corr() to calculate the correlation matrix. Display both matrices using the print() function. The covariance matrix shows how attributes vary together, while the correlation matrix indicates the strength and direction of the linear relationship between each pair of attributes.
