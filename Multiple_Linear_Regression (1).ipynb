```python
#M.PAVANI
#AP24110011675
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score, mean_squared_error

# Example: house price (lakhs) from area, bedrooms and age
df = pd.DataFrame({
    "Area":     [800, 1000, 1200, 1500, 1800, 2000, 2200, 2500, 2800, 3000, 900, 1100, 1400, 1700, 1900, 2100, 2400, 2600, 2900, 3200],
    "Bedrooms": [1, 2, 2, 3, 3, 4, 3, 4, 5, 4, 2, 2, 3, 3, 4, 3, 4, 4, 5, 5],
    "Age":      [10, 8, 12, 5, 6, 3, 9, 2, 4, 1, 11, 7, 6, 8, 4, 5, 3, 6, 2, 1],
    "Price":    [30, 42, 48, 66, 78, 92, 90, 112, 125, 135, 36, 45, 60, 72, 86, 88, 105, 114, 128, 145]
})
print(df.head())

X = df[["Area", "Bedrooms", "Age"]]
Y = df["Price"]
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.25, random_state=1)

model = LinearRegression()
model.fit(X_train, Y_train)
Y_pred = model.predict(X_test)

print("Coefficients:", model.coef_)
print("Intercept:", model.intercept_)
print("R2 score:", r2_score(Y_test, Y_pred))
print("RMSE:", np.sqrt(mean_squared_error(Y_test, Y_pred)))

# Predict for a new house
new_house = pd.DataFrame({"Area": [1600], "Bedrooms": [3], "Age": [5]})
print("Predicted price for new house:", model.predict(new_house)[0])

# Plot: Actual vs Predicted
plt.scatter(Y_test, Y_pred, color="blue")
plt.plot([Y_test.min(), Y_test.max()], [Y_test.min(), Y_test.max()], color="red", label="Perfect prediction")
plt.xlabel("Actual Price")
plt.ylabel("Predicted Price")
plt.title("Multiple Linear Regression - AP24110011675")
plt.legend()
plt.show()
```

Multiple Linear Regression

Uses more than one input feature to predict one output.

Formula: \(Y = b_0 + b_1X_1 + b_2X_2 + b_3X_3\)

Each coefficient shows the change in Y for a one-unit change in that feature, keeping the others constant. Age has a negative coefficient: older houses are cheaper.

Example: Predicting house price from area, bedrooms and age.
