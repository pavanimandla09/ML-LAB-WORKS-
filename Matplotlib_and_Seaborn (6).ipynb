## 1. Loading the Dataset

First step is always to get the data into a DataFrame so we can work with it.
I'm reading the CSV straight from a URL here (instead of downloading it to a file
first) — that way there's only **one cell** to run and no risk of a
`FileNotFoundError` if I run cells out of order or restart the runtime.

```python
import pandas as pd
import numpy as np

# reading the csv directly from the link, no need to download it separately
data = pd.read_csv('https://raw.githubusercontent.com/Aranaur/aranaur.rbind.io/main/datasets/video_games_sales/vgsales.csv')

# just peeking at the first 5 rows to check it loaded fine
data.head()
```

## 2. Importing Matplotlib

`matplotlib.pyplot` is the module we actually use for plotting. Everyone imports it
as `plt`, it's basically the standard.

```python
import matplotlib.pyplot as plt
```

## 3. A Simple Line Plot

Easiest plot to start with — just give `plt.plot()` an x list and a y list.

```python
x_val=[0,1,2]
y_val=[3,5,9]

# x and y have to be the same length, or this will throw an error
plt.plot(x_val, y_val)
```

## 4. Counting Games by Genre

Before I can plot how many games are in each genre, I need the actual counts.
`value_counts()` does this in one line — it counts how many times each unique
value shows up in that column, and sorts them from highest to lowest.

```python
cat_count= data['Genre'].value_counts()
```

```python
# checking what this actually looks like
cat_count
```

```python
# just the genre names, without the counts
cat_count.index
```

## 5. Bar Chart of Genre Counts

Now plotting the same counts as a bar chart. I built it up step by step below —
starting plain, then making the figure bigger, rotating the labels so they don't
overlap, changing the bar width, and finally adding color + a title.

```python
x_bar= cat_count.index
y_bar= cat_count
plt.bar(x_bar, y_bar)
```

```python
# the default size is too small, so let's make the figure bigger
plt.figure(figsize=(12,8))                                                                  #AP24110011675
plt.bar(x_bar, y_bar)
```

```python
#rotate the x labels so they don't overlap each other
plt.figure(figsize=(12,8))
plt.bar(x_bar, y_bar)
plt.xticks(rotation=90, fontsize=12)                                                                       #AP24110011675
```

```python
#Change the width of these bars, thinner looks a bit cleaner here
plt.figure(figsize=(12,8))
plt.bar(x_bar, y_bar, width=0.2)
plt.xticks(rotation=90, fontsize=12)
```

```python
# adding color + proper title and axis labels to finish it off
plt.figure(figsize=(12,8))
plt.bar(x_bar, y_bar, width=0.2, color='orange')
plt.xticks(rotation=90, fontsize=12)
plt.title("Games per Genre", fontsize= 15)
plt.xlabel("Genre")
plt.ylabel("Count of Games")
```

```python
# same as above, just calling plt.show() at the end to display it cleanly
plt.figure(figsize=(12,8))
plt.bar(x_bar, y_bar, width=0.2, color='orange')
plt.xticks(rotation=90, fontsize=12)
plt.title("Games per Genre", fontsize= 15)
plt.xlabel("Genre")
plt.ylabel("Count of Games")
plt.show()
```

## 7. Countplot with Seaborn

Seaborn can skip the `value_counts()` step entirely — `countplot()` counts and
plots in one go, straight from the raw data. Imported it as `sbn` here.

```python
import seaborn as sbn
```

```python
# order=cat_count.index just keeps the bars in the same highest-to-lowest order as before
sbn.countplot(x='Genre', data =data , order= cat_count.index)
plt.show()
```

## 8. Pie Chart of Regional Sales

Switching gears — now looking at how sales are split across regions
(North America, Europe, Japan, and everywhere else) using a pie chart.

```python
#pie chart

# grabbing only the 4 sales columns I actually need for this                                 ##AP24110011675
sales_data=data[["NA_Sales","EU_Sales","JP_Sales","Other_Sales"]]
```

```python
sales_data
```

```python
# .T just flips rows and columns (transpose), makes the next step easier
sales_data.T
```

```python
# summing each region's column to get one total number per region
regional_sale= sales_data.T.sum(axis='columns')
```

```python
regional_sale
```

```python
# autopct shows the percentage on each slice, rounded to 2 decimal places
plt.pie(regional_sale, labels=regional_sale.index, autopct='%.2f%%' )
```

```python
# explode pulls one slice (Europe here) out a bit to highlight it
plt.pie(regional_sale, labels=regional_sale.index, autopct='%.2f%%', explode= (0,0.5,0,0) )
plt.show()                                                                                          #AP24110011675
```

## 9. Histograms

Now checking how the games are spread out over the years using a histogram —
first with the default number of bins, then with more bins to see the trend
year by year more clearly.

```python
#histograms

plt.hist(data['Year'])
plt.show()
```

```python
#popularity of games year on year, more bins = more detail

plt.hist(data['Year'], bins= 50)
plt.show()
```

```python
#seaborn library, does basically the same thing with nicer defaults

sbn.histplot(data['Year'])
```

```python
sbn.histplot(data['Year'], bins=10)
```

```python
sbn.kdeplot(data['Year']) #more smoothened version of histogram, more interpretable
```

## 10. Summary Statistics

```python
data.describe()
```

```python
# same thing but just for one column                                                                                        #AP24110011675
data['NA_Sales'].describe()
```

## 11. Boxplot

Last one — a boxplot of `NA_Sales`. This is good for spotting outliers, since
from `.describe()` above I could already tell the max is way bigger than the
75th percentile, so I expected to see a lot of dots above the box.

```python
plt .figure(figsize=(12,10))
sbn.boxplot(y=data['NA_Sales'])                                                                        #AP24110011675
plt.ylabel("North America Sales")
plt.yticks(fontsize=20)
```
