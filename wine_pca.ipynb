```python
data =raw_data.drop('Type',axis=1)
```

```python
standard_scaler = StandardScaler()
scaled_data = standard_scaler.fit_transform(data)
scaled_data.shape
```

```python
pca_variance = PCA()
pca_variance.fit(scaled_data)

plt.figure(figsize=(20,7))
xi = np.arange(1, 1+scaled_data.shape[1], step=1)
yi = np.cumsum(pca_variance.explained_variance_ratio_)
plt.plot(xi, yi, marker='o', linestyle='--', color='b')
#AP24110011675
plt.ylim(0.0,1.1)
plt.xlabel('Number of Components')
plt.xticks(np.arange(1, 1+scaled_data.shape[1], step=1))
plt.ylabel('Cumulative variance (%)')
plt.title('Explained variance by each component')
plt.axhline(y=1, color='r', linestyle='-')
plt.gca().xaxis.grid(False)
```

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=3)
pca_scaled = pca.fit_transform(scaled_data)   #AP24110011675

pca_sc_df = pd.DataFrame(data = pca_scaled, columns = ['pc1', 'pc2', 'pc3'])
pca_sc_df.head()
```

```python
wcss = []
for i in range(1, 13):
    kmeans = KMeans(n_clusters=i,random_state=0)
    kmeans.fit(pca_sc_df)
    wcss.append(kmeans.inertia_)

plt.figure(figsize=(20,7))
plt.plot(range(1, 13), wcss,color='green')
plt.scatter(range(1,13),wcss,color='red')    #AP24110011675
plt.title('Elbow Method')
plt.xlabel('Number of clusters')
plt.ylabel('WCSS')
plt.show()
```

```python
from sklearn.metrics import silhouette_score
n_clusters = [2,3,4,5,6,7,8,9,10]
clusters_inertia = []
s_scores = []
for n in n_clusters:
    KM_est = KMeans(n_clusters=n, init='k-means++').fit(pca_sc_df)
    clusters_inertia.append(KM_est.inertia_)
    silhouette_avg = silhouette_score(pca_sc_df, KM_est.labels_)
    s_scores.append(silhouette_avg)
                                                                     #AP24110011675
fig, ax = plt.subplots(figsize=(20,7))
ax = sns.lineplot(x=n_clusters, y=s_scores, marker='o', ax=ax)
ax.set_title("Silhouette score method")
ax.set_xlabel("number of clusters")
ax.set_ylabel("Silhouette score")
ax.axvline(3, ls="--", c="red")
plt.show()
```

```python
kmeans_clf = KMeans(n_clusters = 3, init = 'k-means++', max_iter = 300, n_init = 10, random_state = 7)
prediction_kmeans = kmeans_clf.fit_predict(pca_sc_df)
prediction_kmeans        ##AP24110011675
```

```python
plt.figure(figsize=(25,7))
dendrogram_ward = sch.dendrogram(sch.linkage(pca_sc_df, method  = "ward"))
plt.title('Wine type Dendrogram')
plt.xlabel('Wine type')
plt.ylabel('Euclidean distances')
```

```python
agg_clustering = AgglomerativeClustering(n_clusters=3, linkage='ward', metric='euclidean')
prediction_hier = agg_clustering.fit_predict(pca_sc_df)
```

```python
data['Type_actual'] = raw_data['Type']
data['Type_kmeans'] = prediction_kmeans
data['Type_hier'] = prediction_hier
data.head()
```

2. Why might features with larger numerical scales influence PCA more strongly?
PCA identifies directions with maximum variance. Features containing larger numerical values often have larger variances and therefore influence the principal components more strongly.

For example, in the Wine dataset, proline has much larger numerical values than features such as hue.

Without standardization, proline can dominate the PCA calculation.


3. Is standardization always necessary?

No, but it is generally recommended when features have different units or numerical scales.

Standardization helps ensure that features are compared on a similar scale.

If features are already measured on comparable scales, standardization may not be necessary.

3. Does PCA use class labels to create the components?

No.

PCA is an unsupervised dimensionality reduction technique.

It calculates principal components using the input features without considering class labels.

Class labels are used only to color or identify the samples during visualization.

Question 5: Does PCA select existing features?
1. Are the two principal components original Wine dataset columns?

 No.

PC1 and PC2 are new features created by combining the original 13 features mathematically.

They are not existing columns from the dataset.

For example:

$$ PC1=a_1X_1+a_2X_2+\cdots+a_{13}X_{13} $$

Here, \(a_1,a_2,\ldots,a_{13}\) are weights calculated by PCA, and \(X_1,\ldots,X_{13}\) are the standardized input features.

2. What is the difference between feature selection and feature extraction?


Feature selection

1.Selects existing features

2.Original features remain unchanged

3.Example: Selecting alcohol and malic acid

4.Easier to interpret

Feature extraction

1.Creates new features

2.Original features are transformed

3.Example: Creating PC1 and PC2 using PCA

4.Often harder to interpret

3. Why is PCA considered a feature extraction technique?

 PCA is considered a feature extraction technique because it transforms the original features into new principal components.

These components are linear combinations of the original features and capture directions of maximum variance.

PCA does not simply select existing columns; it creates a new representation of the dataset.
